```python
# ============================================================
# SMARTSTOCK AI
# AUTOMATED EXPIRY DETECTION AND SMART INVENTORY MANAGEMENT
# PROCESS ONLY 20 IMAGES
# OCR + IMAGE PREPROCESSING + ML TRAINING
# ============================================================


# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip -q install easyocr scikit-learn joblib


# ============================================================
# IMPORT LIBRARIES
# ============================================================

import os
import zipfile
import cv2
import re
import random
import hashlib
import numpy as np
import pandas as pd
import easyocr
import matplotlib.pyplot as plt
import joblib

from datetime import datetime, date

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report


# ============================================================
# 2. EXTRACT DATASET
# ============================================================

zip_path = "/content/Smart.zip"

extract_path = "/content/smartstock_archive_1222"


print("=" * 60)
print("DATASET COLLECTION")
print("=" * 60)


print(
    "ZIP exists:",
    os.path.exists(zip_path)
)


if not os.path.exists(zip_path):

    raise Exception(
        "Smart.zip not found in /content/"
    )


# Extract ZIP

with zipfile.ZipFile(
    zip_path,
    "r"
) as zip_ref:

    zip_ref.extractall(
        extract_path
    )


print(
    "Archive extracted successfully!"
)


print(
    "Extraction folder:",
    extract_path
)


# ============================================================
# 3. FIND ALL IMAGES INSIDE DATASET
# ============================================================

image_files = []


for root, folders, files in os.walk(
    extract_path
):

    for file in files:

        if file.lower().endswith(
            (
                ".jpg",
                ".jpeg",
                ".png",
                ".jfif",
                ".webp"
            )
        ):

            image_files.append(
                os.path.join(
                    root,
                    file
                )
            )


print(
    "Total images found:",
    len(image_files)
)


if len(image_files) == 0:

    raise Exception(
        "No images found inside Smart.zip"
    )


print("\nFirst 10 images:")


for image in image_files[:10]:

    print(image)


# ============================================================
# 4. DATA CLEANING
# ============================================================

print("\n" + "=" * 60)
print("DATA CLEANING")
print("=" * 60)


valid_images = []

corrupted_images = []

duplicate_images = []


image_hashes = set()


for image_path in image_files:

    image = cv2.imread(
        image_path
    )


    # --------------------------------------------------------
    # CHECK CORRUPTED IMAGE
    # --------------------------------------------------------

    if image is None:

        corrupted_images.append(
            image_path
        )

        continue


    # --------------------------------------------------------
    # CHECK IMAGE DIMENSIONS
    # --------------------------------------------------------

    height, width = image.shape[:2]


    if (
        height == 0
        or width == 0
    ):

        corrupted_images.append(
            image_path
        )

        continue


    # --------------------------------------------------------
    # DUPLICATE DETECTION
    # --------------------------------------------------------

    image_hash = hashlib.md5(
        image.tobytes()
    ).hexdigest()


    if image_hash in image_hashes:

        duplicate_images.append(
            image_path
        )

        continue


    image_hashes.add(
        image_hash
    )


    valid_images.append(
        image_path
    )


print(
    "Original images:",
    len(image_files)
)


print(
    "Valid images:",
    len(valid_images)
)


print(
    "Corrupted images:",
    len(corrupted_images)
)


print(
    "Duplicate images:",
    len(duplicate_images)
)


# ============================================================
# 5. SELECT ONLY 20 IMAGES
# ============================================================

random.seed(42)


selected_images = random.sample(
    valid_images,
    min(
        20,
        len(valid_images)
    )
)


print(
    "\nImages selected:",
    len(selected_images)
)


# ============================================================
# 6. IMAGE PREPROCESSING
# ============================================================

print("\n" + "=" * 60)
print("IMAGE PREPROCESSING")
print("=" * 60)


def preprocess_image(
    image_path
):

    image = cv2.imread(
        image_path
    )


    if image is None:

        return None


    # --------------------------------------------------------
    # RESIZE
    # --------------------------------------------------------

    resized = cv2.resize(
        image,
        (224, 224)
    )


    # --------------------------------------------------------
    # GRAYSCALE
    # --------------------------------------------------------

    gray = cv2.cvtColor(
        resized,
        cv2.COLOR_BGR2GRAY
    )


    # --------------------------------------------------------
    # CONTRAST ENHANCEMENT
    # --------------------------------------------------------

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )


    enhanced = clahe.apply(
        gray
    )


    return (
        resized,
        gray,
        enhanced
    )


print(
    "Resize: 224 x 224"
)


print(
    "Grayscale: Applied"
)


print(
    "Contrast enhancement: Applied"
)


# ============================================================
# 7. DATASET VALIDATION
# ============================================================

print("\n" + "=" * 60)
print("DATASET VALIDATION")
print("=" * 60)


validated_images = []


for image_path in selected_images:

    image = cv2.imread(
        image_path
    )


    if image is None:

        continue


    height, width = image.shape[:2]


    if (
        height < 20
        or width < 20
    ):

        continue


    validated_images.append(
        image_path
    )


print(
    "Selected images:",
    len(selected_images)
)


print(
    "Validated images:",
    len(validated_images)
)


# ============================================================
# 8. LOAD EASY OCR
# ============================================================

print("\n" + "=" * 60)
print("EXPIRY DATE OCR")
print("=" * 60)


reader = easyocr.Reader(
    ["en"],
    gpu=False
)


print(
    "EasyOCR loaded!"
)


# ============================================================
# 9. EXPIRY KEYWORDS
# ============================================================

expiry_keywords = [

    "EXP",
    "EXP.",
    "EXPIRY",
    "EXP DATE",
    "EXPDATE",
    "EXPIRY DATE",
    "BEST BEFORE",
    "USE BY",
    "VALID UNTIL"

]


# ============================================================
# 10. DATE PATTERNS
# ============================================================

date_patterns = [

    # YYYY-MM-DD
    r"\b\d{4}[-/.]\d{1,2}[-/.]\d{1,2}\b",

    # DD-MM-YYYY
    r"\b\d{1,2}[-/.]\d{1,2}[-/.]\d{4}\b",

    # DD-MM-YY
    r"\b\d{1,2}[-/.]\d{1,2}[-/.]\d{2}\b",

    # YYYYMMDD
    r"\b\d{8}\b",

    # YYYY MM DD
    r"\b\d{4}\s+\d{1,2}\s+\d{1,2}\b",

    # DD MM YYYY
    r"\b\d{1,2}\s+\d{1,2}\s+\d{4}\b",

    # DD MAY 2026
    r"\b\d{1,2}\s+"
    r"(JAN|FEB|MAR|APR|MAY|JUN|JUL|AUG|SEP|OCT|NOV|DEC)"
    r"[A-Z]*\s+\d{4}\b"

]


# ============================================================
# 11. FIND DATES
# ============================================================

def find_dates(
    text
):

    text = text.upper()


    found = []


    for pattern in date_patterns:

        matches = re.findall(
            pattern,
            text
        )


        for match in matches:

            if isinstance(
                match,
                tuple
            ):

                match = " ".join(
                    match
                )


            if match not in found:

                found.append(
                    match
                )


    return found


# ============================================================
# 12. CONVERT DATE
# ============================================================

def convert_to_date(
    text
):

    if text is None:

        return None


    text = text.strip().upper()


    formats = [

        "%Y-%m-%d",
        "%Y/%m/%d",
        "%Y.%m.%d",

        "%d-%m-%Y",
        "%d/%m/%Y",
        "%d.%m.%Y",

        "%d-%m-%y",
        "%d/%m/%y",
        "%d.%m.%y",

        "%Y %m %d",
        "%d %m %Y",

        "%d %B %Y",
        "%d %b %Y"

    ]


    for fmt in formats:

        try:

            return datetime.strptime(
                text,
                fmt
            ).date()


        except:

            pass


    # --------------------------------------------------------
    # YYYYMMDD
    # --------------------------------------------------------

    if re.fullmatch(
        r"\d{8}",
        text
    ):

        try:

            return datetime.strptime(
                text,
                "%Y%m%d"
            ).date()


        except:

            pass


    return None


# ============================================================
# 13. STATUS
# ============================================================

def calculate_status(
    expiry_date
):

    if expiry_date is None:

        return (
            "Unknown",
            None
        )


    today = date.today()


    days = (
        expiry_date - today
    ).days


    if days < 0:

        status = "Expired"


    elif days <= 30:

        status = "Near Expiry"


    else:

        status = "Safe"


    return (
        status,
        days
    )


# ============================================================
# 14. OCR + BOUNDING BOX
# ============================================================

def detect_expiry(
    image_path
):

    image = cv2.imread(
        image_path
    )


    if image is None:

        return (
            None,
            "",
            None,
            "Unknown",
            None
        )


    # --------------------------------------------------------
    # RESIZE LARGE IMAGE
    # --------------------------------------------------------

    height, width = image.shape[:2]


    if width > 1200:

        scale = 1200 / width


        image = cv2.resize(
            image,
            None,
            fx=scale,
            fy=scale
        )


    # --------------------------------------------------------
    # GRAYSCALE
    # --------------------------------------------------------

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )


    # --------------------------------------------------------
    # CONTRAST
    # --------------------------------------------------------

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )


    enhanced = clahe.apply(
        gray
    )


    # --------------------------------------------------------
    # EASY OCR
    # --------------------------------------------------------

    results = reader.readtext(
        enhanced,
        detail=1,
        paragraph=False
    )


    detected = []

    all_text = []


    for result in results:

        box = result[0]

        text = result[1]

        confidence = result[2]


        detected.append(
            (
                box,
                text,
                confidence
            )
        )


        all_text.append(
            text
        )


    full_text = " ".join(
        all_text
    ).upper()


    # ========================================================
    # FIND EXPIRY DATE
    # ========================================================

    expiry_text = None

    expiry_box = None


    # --------------------------------------------------------
    # FIRST:
    # KEYWORD + DATE TOGETHER
    # --------------------------------------------------------

    for box, text, confidence in detected:

        upper_text = text.upper()


        dates = find_dates(
            upper_text
        )


        has_keyword = any(
            keyword in upper_text
            for keyword in expiry_keywords
        )


        if (
            has_keyword
            and dates
        ):

            expiry_text = dates[0]

            expiry_box = box

            break


    # --------------------------------------------------------
    # SECOND:
    # KEYWORD + NEARBY DATE
    # --------------------------------------------------------

    if expiry_text is None:

        for i, item in enumerate(
            detected
        ):

            text = item[1]

            upper_text = text.upper()


            has_keyword = any(
                keyword in upper_text
                for keyword in expiry_keywords
            )


            if has_keyword:

                for j in range(
                    max(
                        0,
                        i - 2
                    ),
                    min(
                        len(detected),
                        i + 4
                    )
                ):

                    nearby_text = (
                        detected[j][1]
                    )


                    dates = find_dates(
                        nearby_text
                    )


                    if dates:

                        expiry_text = (
                            dates[0]
                        )


                        expiry_box = (
                            detected[j][0]
                        )


                        break


                if expiry_text:

                    break


    # --------------------------------------------------------
    # THIRD:
    # GENERAL DATE
    # --------------------------------------------------------

    if expiry_text is None:

        dates = find_dates(
            full_text
        )


        if dates:

            expiry_text = dates[0]


    # --------------------------------------------------------
    # CONVERT DATE
    # --------------------------------------------------------

    expiry_date = convert_to_date(
        expiry_text
    )


    # --------------------------------------------------------
    # STATUS
    # --------------------------------------------------------

    status, days = calculate_status(
        expiry_date
    )


    # ========================================================
    # DRAW BOUNDING BOXES
    # ========================================================

    output = image.copy()


    for box, text, confidence in detected:

        points = np.array(
            box,
            dtype=np.int32
        )


        # Blue = OCR text

        cv2.polylines(
            output,
            [points],
            True,
            (255, 0, 0),
            2
        )


    # --------------------------------------------------------
    # GREEN = EXPIRY DATE
    # --------------------------------------------------------

    if expiry_box is not None:

        points = np.array(
            expiry_box,
            dtype=np.int32
        )


        cv2.polylines(
            output,
            [points],
            True,
            (0, 255, 0),
            4
        )


    return (
        output,
        full_text,
        expiry_date,
        status,
        days
    )


# ============================================================
# 15. PROCESS 20 IMAGES
# ============================================================

print("\n" + "=" * 60)
print("PROCESSING 20 IMAGES")
print("=" * 60)


inventory = []


for number, image_path in enumerate(
    validated_images,
    start=1
):


    print(
        f"\nProcessing {number}/"
        f"{len(validated_images)}"
    )


    try:

        (
            output,
            text,
            expiry_date,
            status,
            days
        ) = detect_expiry(
            image_path
        )


        if expiry_date:

            expiry_display = (
                expiry_date.strftime(
                    "%d-%m-%Y"
                )
            )


        else:

            expiry_display = (
                "Not Detected"
            )


        print(
            "Image:",
            os.path.basename(
                image_path
            )
        )


        print(
            "Detected Text:",
            text[:200]
        )


        print(
            "Expiry Date:",
            expiry_display
        )


        print(
            "Status:",
            status
        )


        print(
            "Days Remaining:",
            days
        )


        inventory.append(
            {

                "Image":
                    os.path.basename(
                        image_path
                    ),

                "Detected Text":
                    text[:200],

                "Expiry Date":
                    expiry_display,

                "Status":
                    status,

                "Days Remaining":
                    days
            }
        )


        # ----------------------------------------------------
        # DISPLAY IMAGE
        # ----------------------------------------------------

        plt.figure(
            figsize=(10, 6)
        )


        plt.imshow(
            cv2.cvtColor(
                output,
                cv2.COLOR_BGR2RGB
            )
        )


        plt.title(
            f"Image {number}/20 | "
            f"{status}"
        )


        plt.axis(
            "off"
        )


        plt.show()


    except Exception as e:

        print(
            "Error:",
            e
        )


# ============================================================
# 16. INVENTORY TABLE
# ============================================================

print("\n" + "=" * 60)
print("SMARTSTOCK INVENTORY")
print("=" * 60)


inventory_df = pd.DataFrame(
    inventory
)


display(
    inventory_df
)


# ============================================================
# 17. SAVE OCR RESULTS
# ============================================================

result_file = (
    "/content/smartstock_archive_20_results.csv"
)


inventory_df.to_csv(
    result_file,
    index=False
)


print(
    "\nResults saved:"
)


print(
    result_file
)


# ============================================================
# 18. MACHINE LEARNING DATA PREPARATION
# ============================================================

print("\n" + "=" * 60)
print("ML DATA PREPARATION")
print("=" * 60)


# Remove Unknown images

training_df = inventory_df[
    inventory_df["Status"] != "Unknown"
].copy()


print(
    "Total processed images:",
    len(inventory_df)
)


print(
    "Images available for training:",
    len(training_df)
)


# ============================================================
# 19. SHOW CLASS DISTRIBUTION
# ============================================================

print("\nClass Distribution:")


if len(training_df) > 0:

    print(
        training_df[
            "Status"
        ].value_counts()
    )

else:

    print(
        "No usable training images found."
    )


# ============================================================
# 20. CREATE ML FEATURES
# ============================================================

if len(training_df) > 0:

    X = training_df[
        [
            "Days Remaining"
        ]
    ].copy()


    y = training_df[
        "Status"
    ].copy()


    print(
        "\nTraining Features:"
    )


    print(
        X.head()
    )


# ============================================================
# 21. RANDOM FOREST TRAINING
# ============================================================

if (
    len(training_df) >= 8
    and training_df["Status"].nunique() >= 2
):


    print("\n" + "=" * 60)
    print("RANDOM FOREST MODEL TRAINING")
    print("=" * 60)


    try:

        # ----------------------------------------------------
        # TRAIN / TEST SPLIT
        # ----------------------------------------------------

        X_train, X_test, y_train, y_test = (
            train_test_split(
                X,
                y,
                test_size=0.25,
                random_state=42,
                stratify=y
            )
        )


        print(
            "Training images:",
            len(X_train)
        )


        print(
            "Testing images:",
            len(X_test)
        )


        # ----------------------------------------------------
        # CREATE MODEL
        # ----------------------------------------------------

        model = RandomForestClassifier(
            n_estimators=100,
            random_state=42
        )


        # ----------------------------------------------------
        # TRAIN
        # ----------------------------------------------------

        model.fit(
            X_train,
            y_train
        )


        print(
            "\nModel training completed!"
        )


        # ----------------------------------------------------
        # PREDICTION
        # ----------------------------------------------------

        predictions = model.predict(
            X_test
        )


        # ----------------------------------------------------
        # ACCURACY
        # ----------------------------------------------------

        accuracy = accuracy_score(
            y_test,
            predictions
        )


        print("\n" + "=" * 60)
        print("MODEL EVALUATION")
        print("=" * 60)


        print(
            "Accuracy:",
            round(
                accuracy * 100,
                2
            ),
            "%"
        )


        # ----------------------------------------------------
        # CLASSIFICATION REPORT
        # ----------------------------------------------------

        print(
            "\nClassification Report:"
        )


        print(
            classification_report(
                y_test,
                predictions,
                zero_division=0
            )
        )


        # ----------------------------------------------------
        # SAVE MODEL
        # ----------------------------------------------------

        model_file = (
            "/content/smartstock_model.pkl"
        )


        joblib.dump(
            model,
            model_file
        )


        print(
            "\nTrained model saved:"
        )


        print(
            model_file
        )


    except ValueError as e:

        print(
            "\nModel training error:"
        )


        print(
            e
        )


        print(
            "\nAdd more images to each status class."
        )


else:

    print("\n" + "=" * 60)
    print("MODEL TRAINING SKIPPED")
    print("=" * 60)


    print(
        "Not enough usable data for training."
    )


    print(
        "Unknown images are not used for training."
    )


    print(
        "At least two status classes are required."
    )


# ============================================================
# 22. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)


print(
    "Images found:",
    len(image_files)
)


print(
    "Images selected:",
    len(selected_images)
)


print(
    "Images validated:",
    len(validated_images)
)


print(
    "Images processed:",
    len(inventory_df)
)


if len(inventory_df) > 0:

    print(
        "Safe:",
        sum(
            inventory_df["Status"]
            == "Safe"
        )
    )


    print(
        "Near Expiry:",
        sum(
            inventory_df["Status"]
            == "Near Expiry"
        )
    )


    print(
        "Expired:",
        sum(
            inventory_df["Status"]
            == "Expired"
        )
    )


    print(
        "Unknown:",
        sum(
            inventory_df["Status"]
            == "Unknown"
        )
    )


print(
    "\nSMARTSTOCK AI COMPLETED!"
)
```
